```python
import math


# 1. The point the lights are pointing AT (Your object's centroid)
TARGET_CENTROID = [0.0, 0.0, 7.025] 

# 2. Where should the Radius and Zenith be measured from?
# True  = Measured from the absolute floor/origin [0, 0, 0]
# False = Measured from the TARGET_CENTROID [0, 0, 2.7]
USE_ORIGIN_FOR_ZENITH = True

# 3. Distance from the chosen measurement origin (in cm)
RADIUS_CM = 7.772

# 4. Light Angles in Degrees: (Azimuth, Zenith)
# Azimuth: 0 = +X axis, 90 = +Y axis, 180 = -X axis, 270 = -Y axis
# Zenith: 0 = Straight up (+Z), 90 = Flat on the XY horizon
LIGHT_ANGLES = {
    "L1": {"azimuth": 0.0,   "zenith": 25.328},
    "L2": {"azimuth": 90.0,  "zenith": 60.00},
    "L3": {"azimuth": 180.0, "zenith": 25.328},
    "L4": {"azimuth": 270.0, "zenith": 60.00}
}


# =========================================================
# UTILITIES & MATH
# =========================================================

def spherical_to_cartesian(r, azimuth_deg, zenith_deg, origin):
    """Converts spherical angles to 3D coordinates relative to an origin."""
    azimuth_rad = math.radians(azimuth_deg)
    zenith_rad = math.radians(zenith_deg)

    # Standard Spherical to Cartesian formula (Z-up coordinate system)
    x = r * math.sin(zenith_rad) * math.cos(azimuth_rad)
    y = r * math.sin(zenith_rad) * math.sin(azimuth_rad)
    z = r * math.cos(zenith_rad)

    # Offset by the chosen measurement origin
    return [origin[0] + x, origin[1] + y, origin[2] + z]

def calculate_direction_vector(pos, target):
    """Calculates the normalized direction vector pointing from pos to target."""
    dx = target[0] - pos[0]
    dy = target[1] - pos[1]
    dz = target[2] - pos[2]
    
    # Calculate magnitude (length)
    magnitude = math.sqrt(dx**2 + dy**2 + dz**2)
    
    # Normalize (divide by length)
    return [dx / magnitude, dy / magnitude, dz / magnitude]


# =========================================================
# EXECUTION
# =========================================================

def generate_light_rig():
    
    # Determine the origin based on the user's flag
    if USE_ORIGIN_FOR_ZENITH:
        measurement_origin = [0.0, 0.0, 0.0]
        mode_text = "Absolute Floor [0, 0, 0]"
    else:
        measurement_origin = TARGET_CENTROID
        mode_text = f"Target Centroid {TARGET_CENTROID}"
        
    print(f"--- Generating Rig ---")
    print(f"Radius: {RADIUS_CM} cm | Measured from: {mode_text} | Pointing at: {TARGET_CENTROID}\n")

    calculated_positions = []
    calculated_directions = []

    for name, angles in LIGHT_ANGLES.items():
        # 1. Calculate the 3D Position
        pos = spherical_to_cartesian(
            r=RADIUS_CM, 
            azimuth_deg=angles["azimuth"], 
            zenith_deg=angles["zenith"], 
            origin=measurement_origin
        )
        
        # 2. Calculate the Direction pointing to the target
        direction = calculate_direction_vector(pos, TARGET_CENTROID)
        
        calculated_positions.append(pos)
        calculated_directions.append(direction)

        # Print the results nicely formatted
        print(f"[{name}] Azimuth: {angles['azimuth']}°, Zenith: {angles['zenith']}°")
        print(f"  Pos (cm):  [{pos[0]:.3f}, {pos[1]:.3f}, {pos[2]:.3f}]")
        print(f"  Direction: [{direction[0]:.3f}, {direction[1]:.3f}, {direction[2]:.3f}]\n")

    # Print pure arrays you can easily copy/paste into your Blender script!
    print("--- Copy/Paste Ready Arrays ---")
    print("COORDINATES_CM = [")
    for p in calculated_positions:
        print(f"    [{p[0]:.3f}, {p[1]:.3f}, {p[2]:.3f}],")
    print("]")

    print("\nDIRECTIONS = [")
    for d in calculated_directions:
        print(f"    [{d[0]:.3f}, {d[1]:.3f}, {d[2]:.3f}],")
    print("]")

if __name__ == "__main__":
    generate_light_rig()
```

```python
import math

# =========================================================
# ⚙️ CONFIGURATION
# =========================================================

# 1. The desired Angle of Incidence (theta_in) at the target point
# (This is the zenith angle the light rays will make when they hit the object)
THETA_IN_DEG = 90.0 

# 2. The exact point (X, Y, Z) where Radius (r) and Zenith (phi) are measured FROM
MEASUREMENT_ORIGIN = [0.0, 0.0, 2.7] 

# 3. Distance from the chosen MEASUREMENT_ORIGIN (in cm)
RADIUS_CM = 5.455

# 4. Light Angles in Degrees: (Azimuth, Zenith/Phi)
# Azimuth: 0 = +X axis, 90 = +Y axis, 180 = -X axis, 270 = -Y axis
# Zenith (phi): The angle from the Z-axis, measured from MEASUREMENT_ORIGIN
LIGHT_ANGLES = {
    "L1": {"azimuth": 0.0,   "zenith": 37.55},
    "L2": {"azimuth": 90.0,  "zenith": 37.55},
    "L3": {"azimuth": 180.0, "zenith": 37.55},
    "L4": {"azimuth": 270.0, "zenith": 37.55}
}


# =========================================================
# UTILITIES & MATH
# =========================================================

def solve_target_z(r, phi_deg, theta_in_deg, origin_z):
    """
    Solves for the target Z coordinate using the provided cos(theta_in) equation.
    """
    phi_rad = math.radians(phi_deg)
    theta_in_rad = math.radians(theta_in_deg)
    
    # R = r * sin(phi)
    r_sin_phi = r * math.sin(phi_rad)
    
    # H = r * cos(phi) + 2.7 - z  --->  Rearranged to solve for H based on the angle
    H = r_sin_phi / math.tan(theta_in_rad)
    
    # Solve for z
    z = (r * math.cos(phi_rad) + origin_z) - H
    return z

def spherical_to_cartesian(r, azimuth_deg, zenith_deg, origin):
    """Converts spherical angles to 3D coordinates relative to an origin."""
    azimuth_rad = math.radians(azimuth_deg)
    zenith_rad = math.radians(zenith_deg)

    x = r * math.sin(zenith_rad) * math.cos(azimuth_rad)
    y = r * math.sin(zenith_rad) * math.sin(azimuth_rad)
    z = r * math.cos(zenith_rad)

    return [origin[0] + x, origin[1] + y, origin[2] + z]

def calculate_direction_vector(pos, target):
    """Calculates the normalized direction vector pointing from pos to target."""
    dx = target[0] - pos[0]
    dy = target[1] - pos[1]
    dz = target[2] - pos[2]
    
    magnitude = math.sqrt(dx**2 + dy**2 + dz**2)
    return [dx / magnitude, dy / magnitude, dz / magnitude]


# =========================================================
# EXECUTION
# =========================================================

def generate_light_rig():
    
    # 1. Dynamically calculate the target Z coordinate using L1's phi
    # (Assuming a symmetric rig where all lights have the same r and zenith)
    phi_deg = LIGHT_ANGLES["L1"]["zenith"]
    calculated_z = solve_target_z(RADIUS_CM, phi_deg, THETA_IN_DEG, MEASUREMENT_ORIGIN[2])
    
    target_centroid = [0.0, 0.0, calculated_z]
    
    print(f"--- Generating Rig ---")
    print(f"Radius (r): {RADIUS_CM} cm | Measured from: {MEASUREMENT_ORIGIN}")
    print(f"Desired Theta_in: {THETA_IN_DEG}°")
    print(f"Calculated Target Centroid: [{target_centroid[0]:.3f}, {target_centroid[1]:.3f}, {target_centroid[2]:.3f}]\n")

    calculated_positions = []
    calculated_directions = []

    for name, angles in LIGHT_ANGLES.items():
        # 1. Calculate the 3D Position
        pos = spherical_to_cartesian(
            r=RADIUS_CM, 
            azimuth_deg=angles["azimuth"], 
            zenith_deg=angles["zenith"], 
            origin=MEASUREMENT_ORIGIN
        )
        
        # 2. Calculate the Direction pointing to the newly calculated target
        direction = calculate_direction_vector(pos, target_centroid)
        
        calculated_positions.append(pos)
        calculated_directions.append(direction)

        print(f"[{name}] Azimuth: {angles['azimuth']}°, Zenith (Phi): {angles['zenith']}°")
        print(f"  Pos (cm):  [{pos[0]:.3f}, {pos[1]:.3f}, {pos[2]:.3f}]")
        print(f"  Direction: [{direction[0]:.3f}, {direction[1]:.3f}, {direction[2]:.3f}]\n")

    print("--- Copy/Paste Ready Arrays ---")
    print("COORDINATES_CM = [")
    for p in calculated_positions:
        print(f"    [{p[0]:.3f}, {p[1]:.3f}, {p[2]:.3f}],")
    print("]")

    print("\nDIRECTIONS = [")
    for d in calculated_directions:
        print(f"    [{d[0]:.3f}, {d[1]:.3f}, {d[2]:.3f}],")
    print("]")

if __name__ == "__main__":
    generate_light_rig()
```
