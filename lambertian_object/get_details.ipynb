```python
#set number 2

import pandas as pd

CSV_PATH_Set_2 = r"D:\Chandana\Photometric_Stereo\PROTOTYPE_DESIGN\lambertian_object\histogram_outputs\comparison_analysis\sample_5_3.0x2.1\set_2\comparison_analysis_results.csv"
CSV_PATH_Set_3 = r"D:\Chandana\Photometric_Stereo\PROTOTYPE_DESIGN\lambertian_object\histogram_outputs\comparison_analysis\sample_5_3.0x2.1\set_3\comparison_analysis_results.csv"
df = pd.read_csv(CSV_PATH_Set_2)
#df = pd.read_csv(CSV_PATH_Set_3)


sample_no = 5
orientation = [0.0, 5.0,10.0,15.0,20.0]              
shutter_speed = ["1/13"]   
light_number = 2     


filtered_df = df[
    (df["sample_number"] == sample_no) &
    (df["light_number"]==light_number)&
    df["orientation_angle_deg"].isin(orientation) &
    df["shutter_speed"].isin(shutter_speed)
]

if filtered_df.empty:
    print("\nNo images found for given inputs.")
else:
    print("\nMatching images:")
    cols_to_hide = [
        "aperture",
        "iso",
        "sample_number",
        "shutter_speed",
        "set_number",
        "image_name"
    ]
    print(
        filtered_df
        .drop(columns=cols_to_hide, errors="ignore")
        .reset_index(drop=True)
    )
```
